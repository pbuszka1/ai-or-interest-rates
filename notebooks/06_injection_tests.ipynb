# 6. How sensitive is each design? (injection tests)

**Question:** No method is perfect. If AI had cut employment by X%, would our designs have caught it, or is the effect too small to see?

**How to use:** Run All. It shows the saved results in seconds. To recompute from the data, set `RUN = True` (about 15 minutes; needs the `data/` folder, which isn't in the public repo).

```python
from _nb import *

RUN = False
if RUN:
    run("injection_test")
```

## How it works

- Take control units that had no AI exposure, **plant a fake cut of known size** from 2023, rerun the exact
  analysis, 100 times per size.
- **Power** = how often the planted cut is detected. The **minimum detectable effect** is the smallest size caught
  80% of the time. A "no effect found" only counts if the design could have seen an effect.

## Results

```python
r = results("injection_test")
t = pd.DataFrame(r["rows"]).pivot(index="design", columns="size", values="power")
t.columns = [f"cut {c:.1%}" for c in t.columns]
t.index = t.index.map({"ces": "Industry employment (CES)", "qwi": "Hires by age (QWI)", "cps": "Young vs old (CPS)"})
print("Share of 100 runs where the planted cut was detected (power):")
t
```

## What it means

**The industry design is strong; the age designs are weak.** Industry employment catches a 10% cut 100% of the
time (detectable effect about 5%), so its null result in notebook 3 really rules out large effects. The two age
designs only catch about 40% of 15% cuts (detectable effect about 20–24%), so "no widening after ChatGPT" there is weak
evidence. That's why the young-worker question is still open.
