# 5. Young vs older workers in AI-exposed occupations (CPS)

**Question:** In the most AI-exposed occupations, did 22–27 year olds lose ground relative to 35–65 year olds?

**How to use:** Run All. It shows the saved results in seconds. To recompute from the data, set `RUN = True` (about 3 minutes; needs the `data/` folder, which isn't in the public repo).

```python
from _nb import *

RUN = False
if RUN:
    run("sdid_occupation")
```

## How it works

- **The closest public-data match to the Stanford "Canaries" study.**
- **Units:** occupation groups with enough young CPS respondents (19 high-exposure vs 26 low-exposure).
- **Outcome:** log employment age 22–27 minus age 35–65 in the same occupation.

## Results

```python
r = results("sdid_occupation")
rows = {"2019 baseline (primary)": "primary_2019_baseline", "Late-2021 baseline": "secondary_late2021_baseline",
        "Dot-com bust (no AI)": "dotcom"}
display(pd.DataFrame({k: {"effect": pct(r[v]["high"]["gap"]["tau"]), "95% CI": ci(r[v]["high"]["gap"]["ci95"])}
                      for k, v in rows.items()}).T)
print("Effect by year, pandemic quarters shown:")
by_year(r["exploratory_pandemic_kept"]["high"]["gap"]["by_year"])
```

```python
figure("fig_sdid_occupation")
```

## What it means

**Borderline, and mostly before ChatGPT, with a possible further drop in 2026.** The gap opened in 2021, stayed
around −10 to −14% through 2025, then reached about −23% in 2026 (January–September only). That last step matches
the Stanford direction, but it's one partial year of noisy survey data, and notebook 6 shows this design can't
reliably see effects under about 20%.
